# Conceptos de Reci

Serie de documentos para entender **cómo Reci decide si un residuo es vidrio o
plástico**, desde los conceptos generales de inteligencia artificial hasta las
decisiones concretas tomadas en este repositorio.

Están pensados para leerse en orden. Cada uno asume solo lo anterior y no
requiere conocimientos previos de IA.

## Formato

Todos los documentos son notebooks de Jupyter. Las figuras están **ejecutadas y
embebidas**, así que se ven sin necesidad de correr nada: basta abrir el archivo
en Jupyter, VS Code o GitHub.

Las celdas de código que generan los gráficos quedan visibles a propósito —
puedes cambiar los valores y volver a ejecutarlas para explorar.
